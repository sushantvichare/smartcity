```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px

from IPython.display import display

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
FILE_PATH = r"G:/My Drive/Data Science/smartcity/data/complaints/bmc_complaints.csv" 
df = pd.read_csv(FILE_PATH) 
print("Dataset loaded successfully!") 
print("Shape:", df.shape) 
display(df.head())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
df.info()
display(df.describe(include="all").T)
```

```python
print("Columns:") 
for col in df.columns:
    print(col)
```

```python
# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Remove whitespace from column names
df.columns = df.columns.str.strip()

# Remove leading/trailing spaces from text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()

print("Cleaned columns:")
print(df.columns.tolist())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
missing = (
    df.isna()
    .sum()
    .sort_values(ascending=False)
)

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (
        missing / len(df) * 100
    ).round(2)
})

display(
    missing_df[
        missing_df["Missing_Count"] > 0
    ]
)
```

```python

def find_column(possible_names):

    for name in possible_names:

        for col in df.columns:

            normalized = (
                col.lower()
                .replace(" ", "_")
            )

            if normalized == name:
                return col

    return None


category_col = find_column([
    "category",
    "complaint_category",
    "type",
    "complaint_type"
])

status_col = find_column([
    "status",
    "complaint_status",
    "state"
])

date_col = find_column([
    "date",
    "complaint_date",
    "created_date",
    "created_at",
    "registration_date"
])

print("Category column:", category_col)
print("Status column:", status_col)
print("Date column:", date_col)
```

```python
if date_col:

    df[date_col] = pd.to_datetime(
        df[date_col],
        errors="coerce"
    )

    print("Date datatype:")
    print(df[date_col].dtype)
```

```python
total_complaints = len(df)

print("Total Complaints:", total_complaints)

if status_col:

    status_values = (
        df[status_col]
        .astype(str)
        .str.lower()
    )

    open_count = status_values.str.contains(
        "open",
        na=False
    ).sum()

    resolved_count = status_values.str.contains(
        "resolved|closed|complete",
        na=False
    ).sum()

    pending_count = status_values.str.contains(
        "pending",
        na=False
    ).sum()

    print("Open Complaints:", open_count)
    print("Resolved Complaints:", resolved_count)
    print("Pending Complaints:", pending_count)
```

```python
if category_col:

    category_counts = (
        df[category_col]
        .value_counts()
    )

    display(
        category_counts.to_frame("Count")
    )

if category_col:

    category_counts = (
        df[category_col]
        .value_counts()
        .head(15)
        .sort_values()
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=category_counts.values,
        y=category_counts.index
    )

    plt.title("Top Complaint Categories")
    plt.xlabel("Number of Complaints")
    plt.ylabel("Category")

    plt.show()
```

Complaint Category Pie Chart

```python
if category_col:

    category_counts = (
        df[category_col]
        .value_counts()
        .head(10)
    )

    plt.figure(figsize=(8, 8))

    plt.pie(
        category_counts.values,
        labels=category_counts.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Complaint Distribution by Category")

    plt.show()
```

Complaint Status Distribution

```python
if status_col:

    status_counts = (
        df[status_col]
        .value_counts()
    )

    display(
        status_counts.to_frame("Count")
    )

if status_col:

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        y=status_col,
        order=df[status_col].value_counts().index
    )

    plt.title("Complaint Status Distribution")
    plt.xlabel("Number of Complaints")
    plt.ylabel("Status")

    plt.show()
```

Category vs Status Heatmap

```python
if category_col and status_col:

    category_status = pd.crosstab(
        df[category_col],
        df[status_col]
    )

    display(category_status)

if category_col and status_col:

    plt.figure(figsize=(12, 8))

    sns.heatmap(
        category_status,
        annot=True,
        fmt="d",
        cmap="Blues"
    )

    plt.title(
        "Complaint Category vs Status"
    )

    plt.xlabel("Status")
    plt.ylabel("Category")

    plt.xticks(rotation=45)

    plt.show()
```

```python
## Monthly Complaint Trend
if date_col:

    monthly_complaints = (
        df
        .dropna(subset=[date_col])
        .set_index(date_col)
        .resample("ME")
        .size()
    )

    plt.figure(figsize=(14, 6))

    plt.plot(
        monthly_complaints.index,
        monthly_complaints.values,
        marker="o"
    )

    plt.title(
        "Monthly Complaint Trend"
    )

    plt.xlabel("Month")
    plt.ylabel("Number of Complaints")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()
```

```python
## Monthly Category Heatmap
if date_col and category_col:

    df["month"] = (
        df[date_col]
        .dt.to_period("M")
        .astype(str)
    )

    top_categories = (
        df[category_col]
        .value_counts()
        .head(10)
        .index
    )

    monthly_category = pd.crosstab(
        df["month"],
        df[category_col]
    )

    monthly_category = monthly_category[
        [
            col
            for col in monthly_category.columns
            if col in top_categories
        ]
    ]

    plt.figure(figsize=(14, 8))

    sns.heatmap(
        monthly_category.T,
        cmap="YlOrRd"
    )

    plt.title(
        "Monthly Complaint Category Heatmap"
    )

    plt.xlabel("Month")
    plt.ylabel("Complaint Category")

    plt.show()
```
